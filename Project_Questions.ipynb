```python
class SchoolShop:
    def __init__(self):
        self.books = {
            "Hindi":   {"1-4": 60,  "5-8": 100, "9-10": 150},
            "Marathi":{"1-4": 60,  "5-8": 100, "9-10": 150},
            "English":{"1-4": 80,  "5-8": 100, "9-10": 150},
            "Science":{"1-4": 90,  "5-8": 120, "9-10": 200},
            "Maths":  {"1-4":100,  "5-8": 140, "9-10": 250}
        }

        self.notebooks = {
            "square": {100: 40, 200: 70},
            "4lines": {100: 30, 200: 50},
            "2lines": {100: 30, 200: 50},
            "single": {100: 60, 200: 100},
            "A4":     {100:100, 200:180}
        }

        self.total = 0
        self.bill_items = []
        self.standard_key = ""

    def get_class(self):
        std = input("Enter class (1-10): ")
        if std in ["1","2","3","4"]:
            self.standard_key = "1-4"
        elif std in ["5","6","7","8"]:
            self.standard_key = "5-8"
        elif std in ["9","10"]:
            self.standard_key = "9-10"
        else:
            print("Invalid class")
            exit()

    def buy_book(self):
        try:
            book = input("Enter book name: ")
            qty = int(input("Enter quantity: "))
            price = self.books[book][self.standard_key]
            cost = price * qty
            self.total += cost
            self.bill_items.append(f"{book} Book x{qty} = {cost}")
            print("Book added")
        except:
            print("Invalid book or quantity")

    def buy_notebook(self):
        try:
            nb = input("Enter notebook type: ")
            pages = int(input("Pages (100 / 200): "))
            qty = int(input("Enter quantity: "))
            price = self.notebooks[nb][pages]
            cost = price * qty
            self.total += cost
            self.bill_items.append(f"{nb} Notebook ({pages}) x{qty} = {cost}")
            print("Notebook added")
        except:
            print("Invalid notebook details")

    def show_bill(self):
        print("------ SCHOOL SHOP BILL ------")
        for item in self.bill_items:
            print(item)
        print("Total Amount:", self.total)
        print("Thank you for visiting")


shop = SchoolShop()
shop.get_class()

while True:
    print("1. Buy Book")
    print("2. Buy Notebook")
    print("3. Exit & Print Bill")

    choice = input("Enter choice: ")

    if choice == "1":
        shop.buy_book()
    elif choice == "2":
        shop.buy_notebook()
    elif choice == "3":
        shop.show_bill()
        break
    else:
        print("Invalid option")
```

```python
class EligibilityCheck:
    def __init__(self, dob_year, eligibility_age):
        self.dob_year = dob_year
        self.eligibility_age = eligibility_age

    def check_eligibility(self):
        current_year = 2026
        age = current_year - self.dob_year

        if age >= self.eligibility_age:
            return "Eligible"
        else:
            return "Not Eligible"


try:
    dob_year = int(input("Enter birth year: "))
    eligibility_age = int(input("Enter eligibility age: "))

    obj = EligibilityCheck(dob_year, eligibility_age)
    result = obj.check_eligibility()
    print(result)

except:
    print("Invalid input")
```
