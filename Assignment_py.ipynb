```python
print("="*40)
print("{:^40}".format("BHAKTI DA DHABA"))
print("="*40)

print("{:<3} {:<15} {:<5} {:<7}".format("SR", "ITEM", "QTY", "PRICE"))

total = 0
n = int(input("How many items? "))

bill = []

for i in range(1, n+1):
    name = input("Item name: ")
    qty = int(input("Quantity: "))
    price = int(input("Price: "))

    amt = qty * price
    total += amt

    bill.append((i, name, qty, amt))   # sirf store

print("="*40)


for i in bill:
    print("{:<3} {:<15} {:<5} {:<7}".format(i[0], i[1], i[2], i[3]))

print("="*40)
print("{:<25} {}".format("TOTAL AMOUNT", total))
print("="*40)
```

```python
listt = [
    {"key": {"subkey": 20}},
    {"k2": {"sub2": 5}},
    {"k3": {"sub4": 16}},
    {"k4": {"sub4": 6}}
]

for i in range(len(listt)):
    for j in range(i + 1, len(listt)):

        a = list(listt[i].values())[0]
        a_val = list(a.values())[0]

        b = list(listt[j].values())[0]
        b_val = list(b.values())[0]

        if a_val > b_val:
            listt[i], listt[j] = listt[j], listt[i]

print(listt)
```

```python
from datetime import datetime

start_date = input("Enter FD start date (dd-mm-yyyy): ")
end_date = input("Enter FD end date (dd-mm-yyyy): ")

s_date = datetime.strptime(start_date, "%d-%m-%Y")
e_date = datetime.strptime(end_date, "%d-%m-%Y")

days = (e_date - s_date).days

if days <= 180:
    print("Interest Rate: 4%")
elif days <= 365:
    print("Interest Rate: 5.5%")
elif days <= 1095:
    print("Interest Rate: 6.5%")
else:
    print("Interest Rate: 7%")
```

```python
books = {
    "English": 200,
    "Maths": 250,
    "Science": 300
}

notebook_price = 50

print("----- SCHOOL BOOK BILL -----")

name = input("Enter student name: ")
subject = input("Enter subject (English/Maths/Science): ")

total = 0
if subject in books:
    print(subject, "Book :", books[subject])
    total += books[subject]
else:
    print("Invalid subject")
    exit()

need_notebook = input("Do you need notebook? (yes/no): ")

if need_notebook.lower() == "yes":
    print("Notebook :", notebook_price)
    total += notebook_price
    print("Exit after notebook")

print("----------------------------")
print("Student Name :", name)
print("Total Amount :", total)
print("----------------------------")
```

```python
# Data
books = {
    "Hindi":   {"1-4": 60,  "5-8": 100, "9-10": 150},
    "Marathi":{"1-4": 60,  "5-8": 100, "9-10": 150},
    "English":{"1-4": 80,  "5-8": 100, "9-10": 150},
    "Science":{"1-4": 90,  "5-8": 120, "9-10": 200},
    "Maths":  {"1-4":100,  "5-8": 140, "9-10": 250}
}

notebooks = {
    "square": {100: 40, 200: 70},
    "4lines": {100: 30, 200: 50},
    "2lines": {100: 30, 200: 50},
    "single": {100: 60, 200: 100},
    "A4":     {100:100, 200:180}
}


total = 0
bill_items = []


std_input = input("Enter class (1-10): ").lower()

if "1" in std_input or "2" in std_input or "3" in std_input or "4" in std_input:
    standard_key = "1-4"
elif "5" in std_input or "6" in std_input or "7" in std_input or "8" in std_input:
    standard_key = "5-8"
elif "9" in std_input or "10" in std_input:
    standard_key = "9-10"
else:
    print("Invalid class")
    exit()


while True:
    print("\n1. Buy Book")
    print("2. Buy Notebook")
    print("3. Exit & Print Bill")

    choice = input("Enter choice: ")

    if choice == "1":
        book = input("Enter book name: ")
        qty = int(input("Enter quantity: "))
        price = books[book][standard_key]
        cost = price * qty
        total += cost
        bill_items.append(f"{book} Book x{qty} = {cost}")
        print("Book added")

    elif choice == "2":
        nb = input("Enter notebook type: ")
        pages = int(input("Pages (100 / 200): "))
        qty = int(input("Enter quantity: "))
        price = notebooks[nb][pages]
        cost = price * qty
        total += cost
        bill_items.append(f"{nb} Notebook ({pages}) x{qty} = {cost}")
        print("Notebook added")

    elif choice == "3":
        break

    else:
        print("Invalid option")

print("\n------ SCHOOL SHOP BILL ------")
print("Class:", std_input)
print("------------------------------")

for item in bill_items:
    print(item)

print("------------------------------")
print("Total Amount:", total)
print("------------------------------")
print("Thank you for visiting")
```
